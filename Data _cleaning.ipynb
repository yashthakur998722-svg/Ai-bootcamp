```python
import pandas as pd
import numpy as np

df = pd.read_csv("Book1.csv", encoding="latin1", on_bad_lines="skip")

print("Original Data")
print(df)

print("\nColumns")
print(df.columns)

print("\nMissing Values")
print(df.isnull().sum())

print("\nRows with missing values")
print(df[df.isnull().any(axis=1)])

clean = df.copy()

if "name" in clean.columns:
    clean["name"] = clean["name"].fillna("Unknown")
    clean["name"] = clean["name"].astype(str).str.strip()
    clean["name"] = clean["name"].str.title()

if "age" in clean.columns:
    clean["age"] = pd.to_numeric(clean["age"], errors="coerce")
    clean["age"] = clean["age"].fillna(clean["age"].median())
    clean.loc[clean["age"] < 0, "age"] = np.nan

if "salary" in clean.columns:
    clean["salary"] = clean["salary"].astype(str)
    clean["salary"] = clean["salary"].str.replace(",", "", regex=False)
    clean["salary"] = pd.to_numeric(clean["salary"], errors="coerce")
    clean["salary"] = clean["salary"].fillna(clean["salary"].mean())

if "employee_id" in clean.columns:
    print("\nDuplicate Records")
    print(clean[clean.duplicated(subset="employee_id", keep=False)])

    clean = clean.drop_duplicates(subset="employee_id")


if "city" in clean.columns:
    clean["city"] = clean["city"].astype(str)
    clean["city"] = clean["city"].str.strip()
    clean["city"] = clean["city"].str.title()


if "phone" in clean.columns:
    clean["phone"] = clean["phone"].astype(str)

    clean["phone"] = clean["phone"].str.replace("-", "", regex=False)
    clean["phone"] = clean["phone"].str.replace("(", "", regex=False)
    clean["phone"] = clean["phone"].str.replace(")", "", regex=False)
    clean["phone"] = clean["phone"].str.replace("+91", "", regex=False)

    clean.loc[clean["phone"].str.contains("[A-Za-z]", na=False), "phone"] = np.nan


if "email" in clean.columns:
    clean.loc[~clean["email"].str.contains("@", na=False), "email"] = np.nan


if "join_date" in clean.columns:
    clean["join_date"] = pd.to_datetime(
        clean["join_date"],
        errors="coerce",
        format="mixed"
    )


clean = clean.reset_index(drop=True)

print("\nCleaned Data")
print(clean)

clean.to_csv("clean_employee_data.csv", index=False)

df
```
